```python
def parse_uniprot_flatfile(filepath):
    records = []
    with open(filepath, "r") as f:
        current_record = {}
        for line in f:
            line = line.strip()

            if line.startswith("ID"):
                current_record = {}

            elif line.startswith("AC"):
                # Primary accession is the first one listed before the first semicolon
                accession = line.split(";")[0].replace("AC", "").strip()
                current_record["Accession"] = accession

            elif line.startswith("DE") and "RecName: Full=" in line:
                protein_name = line.split("RecName: Full=")[-1].split("{")[0].strip(" ;")
                current_record["Protein_Name"] = protein_name

            elif line.startswith("GN") and "Name=" in line:
                gene_name = line.split("Name=")[-1].split(" ")[0].strip(" ;")
                current_record["Gene_Name"] = gene_name

            elif line.startswith("OS"):
                organism = line.replace("OS", "").strip(" .")
                current_record["Organism"] = organism

            elif line.startswith("//"):
                if (
                    current_record.get("Accession")
                    and current_record.get("Protein_Name")
                    and current_record.get("Gene_Name")
                    and current_record.get("Organism")
                ):
                    records.append(current_record)

    return records
```

```python
filepath = "/Users/zheng/Downloads/uniprot_sprot.dat"
parsed_records = parse_uniprot_flatfile(filepath)

import pandas as pd
df = pd.DataFrame(parsed_records)
df.head()
```

```python
output_path = "/Users/zheng/Downloads/uniprot_parsed.csv"
df.to_csv(output_path, index=False)
print(f"Saved to: {output_path}")
```
